```python
rng = np.random.default_rng(10)
# seconds to complete checkout; two shoppers in the new flow got stuck on a payment page
old = rng.lognormal(mean=3.0, sigma=0.35, size=40).round(1)
new = np.concatenate([rng.lognormal(mean=3.0, sigma=0.35, size=10), [61.0, 74.5]]).round(1)
print(f"old: mean {old.mean():.1f} s (n={old.size}); new: mean {new.mean():.1f} s (n={new.size})")
```

A two-sample t-test on the checkout times:

```python
%%time
res = stats.ttest_ind(new, old)
print(f"t = {res.statistic:.2f}, p = {res.pvalue:.4f}")
```

```python
display("The new checkout changes completion time" if res.pvalue < 0.05 else "No difference")
```
